```python
w_cherry = [442, 8, 2]
```

```python
w_digital = [5, 1683, 1670]
```

```python
w_information = [5, 3982, 3325]
```

```python
def dot_product(a, b):
    result = 0
    return sum([a_i * b_i for a_i, b_i in zip(a, b)])
```

```python
dot_product(w_cherry, w_digital)
```

```python
def norm(a):
    return math.sqrt(dot_product(a, a))
```

```python
def cosine_similarity(a, b):
    return dot_product(a, b) / (norm(a) * norm(b))
```

```python
cosine_similarity(w_cherry, w_digital)
```

```python
cosine_similarity(w_information, w_digital)
```
