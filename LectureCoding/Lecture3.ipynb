```python
from collections import Counter
```

```python
def tokenize(text):
    return text.lower().split()
```

```python
def train(data):
    counter = Counter()
    for text in data.splitlines():
        tokens = tokenize(text)
        for token in tokens:
            counter[token] += 1

        for t1, t2 in zip(tokens, tokens[1:]):
            counter[(t1, t2)] += 1

    return counter
```

```python
def probability(text, model):
    probs = []
    tokens = tokenize(text)
    for t1, t2 in zip(tokens, tokens[1:]):
        prob = model[(t1, t2)] / model[t1]
        probs.append(prob)

    return probs
```

```python
if __name__ == "__main__":
    data = """I like to eat apples and bananas.
    I like to eat oranges and grapes.
    I like to eat strawberries and blueberries.
    I am sleeping at the moment."""

    counter = train(data)
    print(counter)
```
